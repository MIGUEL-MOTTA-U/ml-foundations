# Getting started with MatplotLib
Matplotlib is a Python library used for creating static, interactive, and animated visualizations such as plots, charts, and graphs.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
s = [i+1 for i in range(20)]
plt.plot(s)

ax = plt.subplot()
ax.set_title("Simple Plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_xlim(-1, 21)
ax.set_ylim(-1, 21)
ax.grid(True)

plt.show()
```

We can create a plot from two lists

```python
l1 = [ i for i in range(21)]
l2 = [ i ** 2 for i in l1]
plt.plot(l1, l2)
plt.show()
```

```python
# We can also take advantage of numpy for this calculus

#help(np.linspace)
x = np.linspace(1,100, 100)
y = np.random.randint(0, 20, 100)
plt.plot(x, y)
ax = plt.subplot()
ax.set_title("Random Plot")
ax.set_xlabel("x") # plt.xlabel("x)
ax.set_ylabel("y") # plt.ylabel("y")
ax.set_xlim(0, 100) # plt.ylim(0, 100)
ax.set_ylim(-1, 21) # plt.xlim(-1, 21)
plt.show()
```

We can use different functions in the same plot

```python
y1 = np.random.randint(0, 20, 50)
y2 = np.random.randint(0, 20, 50)
x = np.linspace(1, 50, 50)
plt.plot(x, y1, x, y2)
plt.show() # This shows the normal graph
```

But we can customize the plot when we create it

```python
plt.plot(x, y1, 'b--', x, y2, 'g-', x, y1/2+y2, 'rx') # '[color][mode]'
plt.show()
```

We can do it in different sentences

```python
plt.plot(x, y1, 'b--', label='y1')
plt.plot(x, y2, 'g-', label='y2')
plt.plot(x, y1/2+y2, 'rx', label="y3") # '[color][mode]'
plt.legend(loc='best')
plt.show()
```

### Subplots

In Matplotlib, a subplot refers to an individual plot or graph that is part of a larger figure containing multiple plots arranged in a grid. Essentially, it allows you to display several related visualizations within a single figure window for easy comparison or organization.

Each subplot has its own coordinate system, axes, and labels, but they share the overall figure space. You can control the layout (number of rows and columns), spacing, and appearance of each subplot to create complex, multi-panel visualizations—such as comparing datasets, showing different variables side by side, or presenting different stages of analysis in one figure.

```python
# We must set the number of rows and columns of the grid where the plot is set, and the number where it is located
y1 = np.random.randint(0, 20, 50)
y2 = np.random.randint(0, 20, 50)
y3 = (y1 / 2) + y2
x = np.linspace(1, 50, 50)

plt.subplot(2,2, 1)
plt.plot(x, y1, 'y-', label='y1')
plt.subplot(2,2, 2)
plt.plot(x, y2, 'b', label='y2')
plt.subplot(2,2, 3)
plt.plot(x, y3, 'g-', label='y3')
plt.show()
```

### Scatter plots

Scatter plots in Matplotlib are a type of data visualization used to display the relationship between two numerical variables. Each point on the plot represents an observation in the dataset, with its position determined by values on the x- and y-axes.

In Matplotlib, scatter plots are typically created to identify patterns, trends, or correlations between variables. They can also show clustering, outliers, and the distribution of data. Points can be customized by color, size, and shape to represent additional dimensions or categories of data.


```python
x, y = np.random.randint(0, 20, 50), np.random.randint(0, 20, 50)
x2, y2 = np.random.randint(0, 20, 50), np.random.randint(0, 20, 50)
```

```python
plt.subplot(1, 2, 1)
plt.subplot(121) # It's the same
plt.scatter(x,y, color='green')
plt.scatter(x2,y2, color='blue')

plt.xlabel("x", fontsize=10, fontweight='bold')
plt.ylabel("y", fontsize=10, fontweight='bold')

plt.ylim(-1, 20)
plt.xlim(-1, 20)

plt.show()
```

### Histograms
**Histograms in Matplotlib** are graphical representations used to show the distribution of a dataset by grouping continuous data into intervals, called *bins*. The height of each bar indicates the frequency (or count) of data points within that range.

In **Matplotlib**, histograms help visualize how data are spread, identify patterns such as skewness or modality (e.g., unimodal, bimodal), and detect outliers. They are commonly used to understand the underlying probability distribution of a variable.


```python
data = np.random.randint(1, 30, 100)
```

```python
plt.subplot(121)
plt.hist(data, bins=30, rwidth=0.8)
# Is the same plot, but if we increase the size of the allowed columns more than the number of points, it'll just reduce the width of the columns
plt.subplot(122)
plt.hist(data, bins=90, rwidth=0.8)
plt.show()
```

### Save figures
